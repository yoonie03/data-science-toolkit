```python
import pandas as pd # library for data frame
import numpy as np # library/package for dealing with arrays/numbers
```

1. Create a numpy array of size 2 x 2 (2 rows and 2 columns)

```python
arr = np.array([[5,7], [0,9]])
print(arr)
```

2. Create a random numpy array of size 3 x 4



```python
rand_arr = np.random.rand(3,4)
print(rand_arr)
```

3. Create a numpy array of ones and a numpy array of zeroes

```python
ones = np.ones((10,5))
zeros = np.zeros((6,6))

print(ones)
print(zeros)
```

4.Create a random numpy array of size 6x3 and transpose it

```python
rand_arr = np.random.rand(6,3)

rand_arr_trans = rand_arr.T
print(rand_arr)
print(rand_arr_trans)
```

5. Create two random numpy array of size n x p and p x m and then compute the dot product

```python
n = 2
p = 3
m = 4

arr1 = np.random.random((n, p))
arr2 = np.random.random((p, m))

arr12 = np.dot(arr1, arr2)

print(arr1)
print(arr2)
print(arr12)
```
