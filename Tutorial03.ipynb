### 8. Does it have missing values? If yes, which column, which row.

While checking the null values in the previous step, I noticed that the `smoking_status` column contains `Unknown` values. So, I searched for a function to check all the unique values in each column and found the `unique()` function. I used it to check what values each column contains.

The `bmi` column has 201 null values. Also, the `smoking_status` column contains `Unknown` values. Although these are not counted as null values, they can be considered missing values because the smoking information is not available. Therefore, both columns contain missing values.

```python
df['smoking_status'].unique()
```

```python
i=1
for col in df.columns:
    print("Column", i, ":", col)
    print("Unique values:", df[col].unique())
    print()
    i+=1
```
